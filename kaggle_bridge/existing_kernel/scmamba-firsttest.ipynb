```python
import os, json, glob, inspect, torch

# --- Checkpoint dosyalarını bul ---
ckpt = glob.glob("/kaggle/input/**/best_model.pt", recursive=True)[0]
args_p = glob.glob("/kaggle/input/**/args.json", recursive=True)[0]
vocab_p = glob.glob("/kaggle/input/**/vocab.json", recursive=True)[0]
print("checkpoint:", ckpt)
print("args     :", args_p)
print("vocab    :", vocab_p)

# --- 1) args.json (model config) ---
print("\n=== args.json ===")
with open(args_p) as f:
    args = json.load(f)
print(json.dumps(args, indent=2)[:2000])

# --- 2) vocab boyutu ---
with open(vocab_p) as f:
    vocab = json.load(f)
print("\nvocab size:", len(vocab), "| ilk 5:", list(vocab.items())[:5])

# --- 3) checkpoint yapısı ---
print("\n=== best_model.pt ===")
sd = torch.load(ckpt, map_location="cpu")
if isinstance(sd, dict) and all(not hasattr(v,"shape") for v in list(sd.values())[:3]):
    print("üst seviye anahtarlar:", list(sd.keys())[:10])
    # state_dict iç anahtarı olabilir
    for k in ("model","state_dict","model_state_dict"):
        if k in sd:
            sd = sd[k]; print(f"-> '{k}' içindeki state_dict kullanılıyor"); break
print("toplam parametre tensörü:", len(sd))
for k in list(sd.keys())[:15]:
    print("   ", k, tuple(sd[k].shape))

# --- 4) model sınıfı imzaları ---
print("\n=== scmamba2.model sınıfları ===")
from scmamba2 import model
for n in dir(model):
    o = getattr(model, n)
    if inspect.isclass(o) and not n.startswith("_"):
        try: print(f"class {n}{inspect.signature(o.__init__)}")
        except Exception as e: print(f"class {n} (?: {e})")

# --- 5) README'de çalıştırma talimatı ---
rd = glob.glob("/kaggle/input/**/README.md", recursive=True)
if rd:
    print("\n=== README (ilk 3000 karakter) ===")
    print(open(rd[0]).read()[:3000])
```

```python
import inspect, glob, os
from scmamba2 import model

# 1) MambaModel.__init__ ve forward kaynak kodu
print("="*90, "\nMambaModel.__init__\n", "="*90)
print(inspect.getsource(model.MambaModel.__init__))
print("="*90, "\nMambaModel.forward\n", "="*90)
print(inspect.getsource(model.MambaModel.forward))

# 2) embedding üreten metodlar (cell emb için)
print("="*90, "\nMambaModel metodları:")
print([m for m in dir(model.MambaModel) if not m.startswith("__")])
for m in dir(model.MambaModel):
    if "emb" in m.lower() or "encode" in m.lower():
        try:
            print("\n--- def", m, "---")
            print(inspect.getsource(getattr(model.MambaModel, m)))
        except Exception as e:
            print(m, "kaynak yok:", e)

# 3) Repoda MambaModel'i KURUP checkpoint YÜKLEYEN örnek scriptleri bul
print("="*90, "\nMambaModel( kullanan / best_model yükleyen scriptler:")
for py in glob.glob("/kaggle/input/**/*.py", recursive=True):
    try:
        txt = open(py).read()
    except: 
        continue
    if "MambaModel(" in txt or "best_model" in txt or "load_state_dict" in txt:
        print("\n>>>", py)
        for i, line in enumerate(txt.splitlines(), 1):
            if any(k in line for k in ["MambaModel(", "load_state_dict", "best_model", "torch.load", "vocab", "cell_emb", "get_batch_cell"]):
                print(f"   {i}: {line.strip()}")
```

```python
import warnings; warnings.simplefilter("ignore")
import json, re, torch, numpy as np
from pathlib import Path
from scmamba2.model import MambaModel
from scmamba2.tokenizer.gene_tokenizer import GeneVocab

CKPT_DIR = Path("/kaggle/input/datasets/tughankovucu/sc-mamba2-checkpoint")
device = "cuda"

# --- config ve vocab ---
cfg   = json.load(open(CKPT_DIR / "args.json"))
vocab = GeneVocab.from_file(CKPT_DIR / "vocab.json")
for tok in ["<pad>", "<cls>", "<eoc>"]:
    if tok not in vocab:
        vocab.append_token(tok)
vocab.set_default_index(vocab["<pad>"])
ntoken = len(vocab)
print("vocab:", ntoken, "| pad idx:", vocab["<pad>"])

# --- checkpoint'i oku, mamba katman sayısını checkpoint'ten türet ---
sd = torch.load(CKPT_DIR / "best_model.pt", map_location="cpu")
if "model" in sd and isinstance(sd["model"], dict): sd = sd["model"]
n_mamba = max(int(m.group(1)) for k in sd
              if (m := re.match(r"transformer_encoder\.layers\.(\d+)\.", k))) + 1
print("checkpoint mamba katman sayisi:", n_mamba)

# --- modeli args.json ile kur ---
model = MambaModel(
    ntoken=ntoken,
    d_model=cfg["embsize"],
    mamba_layer=n_mamba,
    d_hid=cfg["d_hid"],
    nlayers=cfg["nlayers"],
    nlayers_cls=cfg["n_layers_cls"],
    n_cls=1,
    vocab=vocab,
    dropout=cfg["dropout"],
    pad_token=cfg["pad_token"],
    pad_value=cfg["pad_value"],
    input_emb_style=cfg["input_emb_style"],
    n_input_bins=cfg["n_bins"],
    cell_emb_style="cls",
).to(device)

# --- anahtar isimlerini otomatik eşle (transformer_encoder -> mamba_encoder) ---
model_sd = model.state_dict()
remapped = {}
for k, v in sd.items():
    nk = k if k in model_sd else k.replace("transformer_encoder.", "mamba_encoder.")
    remapped[nk] = v
missing, unexpected = model.load_state_dict(remapped, strict=False)
print(f"\nyuklendi | eslesen: {len(remapped)-len(unexpected)} / {len(model_sd)}")
print("missing (ilk 5):", missing[:5])
print("unexpected (ilk 5):", unexpected[:5])

# --- SENTETIK FORWARD (gercek veri olmadan modelin calistigini kanitla) ---
model.eval()
B, L = 2, 512
src  = torch.randint(0, ntoken, (B, L), device=device)
src[:, 0] = vocab["<cls>"]                       # cls token ilk pozisyonda
vals = torch.rand(B, L, device=device) * cfg["n_bins"]
mask = torch.zeros(B, L, dtype=torch.bool, device=device)  # padding yok

with torch.no_grad(), torch.autocast("cuda", dtype=torch.float16):
    out = model(src, vals, mask, CLS=False, MVC=False, ECS=False)

print("\n=== ÇIKTI ===")
print("cell_emb shape :", out["cell_emb"].shape, out["cell_emb"].dtype)
print("mlm_output shape:", out["mlm_output"].shape)
print("cell_emb[0][:8]:", out["cell_emb"][0][:8].float().cpu().numpy())
print("\n✅ MODEL ÇALIŞIYOR — checkpoint yüklendi, forward pass başarılı.")
```
